```python
from jax import grad, jvp, jit
import jax.numpy as jnp
```

```python
class Tracer():
    
    def __init__(self, f):
        self._f = jit(f)
        self._grad = jit(grad(f))
        
        self.counter_f = 0
        self.counter_grad_f = 0
        
    def __call__(self, x):
        self.counter_f += 1
        return self._f(x)
    
    def grad(self, x):
        self.counter_grad_f += 1
        return self._grad(x)
    
    def reset(self):
        self.counter_f = 0
        self.counter_grad_f = 0
    
    def print_info(self):
        print(f"evals: {self.counter_f}\n"
              f"grad evals: {self.counter_grad_f}")
```

```python
a = 20

F = Tracer(lambda x: a*x[0]**2 + x[1]**2)

#F(jnp.array([1.,1.]))

#F.print_info()

F.grad(jnp.array([1.,1.]))

F.print_info()
```

```python
maxiter = 2000
tol = 1e-12
stepsize = 1/80  # insert alpha=1/L, alpha=2/L or alpha=1/(2L) here

x = jnp.array([1.,1.])
F.reset()

for i in range(maxiter):
    
    d = -F.grad(x)
    res_norm = jnp.linalg.norm(d)
    
    if res_norm <= tol:
        break
    elif i % 100 == 0:
        print(f"iteration {i}, ||grad(F)|| = {res_norm}")
        
    x = x + stepsize*d
    
F.print_info()
```

```python
def armijo(F, x, d, slope, c=0.5, tau=0.5, alpha0=1):
    """
    """

    alpha = alpha0
    Fx = F(x)
    cslope = c*slope
    
    while F(x + alpha*d) > Fx + alpha*cslope:
        
        alpha *= tau
        
    return alpha
```

```python
x = jnp.array([1.,1.])
F.reset()

alpha = 1

for i in range(maxiter):
    d = -F.grad(x)
    
    res_norm = jnp.linalg.norm(d)
    
    if res_norm < tol:
        break
    elif i % 100 == 0:
        print(f"iteration {i}, ||grad(F)|| = {res_norm}")
     
    slope = -res_norm**2 
    alpha = armijo(F, x, d, slope, alpha0=1.2*alpha)
    
    x = x + alpha*d
    
F.print_info()
```

```python
# Heavy-ball armijo
x = jnp.array([1.,1.])
y = x.copy()

F.reset()

alpha = 1
beta = 0.1

for i in range(maxiter):
    
    d = -F.grad(x)
    
    res_norm = jnp.linalg.norm(d)
    
    if res_norm < tol:
        break
    elif i % 100 == 0:
        print(f"iteration {i}, ||grad(F)|| = {res_norm}")
     
    slope = F.grad(y)@d
    
    # check if d is a descent direction
    if slope < 0:
        alpha = armijo(F, y, d, slope, alpha0=1.2*alpha)
    else:
        alpha = 0
    
    x_old = x
    #print(f"F(y) = {F(y)}")
    #print(f'alpha = {alpha}')
    x = y + alpha*d
    #print(f"F(x) = {F(x)}")
    y = x + beta*(x - x_old)
    #print(f"F(y) = {F(y)}")
    #print()
    
F.print_info()
```

```python
# Heavy-ball constant
x = jnp.array([1.,1.])
y = x.copy()

F.reset()

alpha = 1
beta = 0.1

for i in range(maxiter):
    
    d = -F.grad(x)
    
    res_norm = jnp.linalg.norm(d)
    
    if res_norm < tol:
        break
    elif i % 100 == 0:
        print(f"iteration {i}, ||grad(F)|| = {res_norm}")

    alpha = 1/20
    
    x_old = x
    #print(f"F(y) = {F(y)}")
    #print(f'alpha = {alpha}')
    x = y + alpha*d
    #print(f"F(x) = {F(x)}")
    y = x + beta*(x - x_old)
    #print(f"F(y) = {F(y)}")
    #print()
    
F.print_info()
```

```python
# Nesterov armijo
x = jnp.array([1.,1.])
y = x.copy()

F.reset()

alpha = 1
beta = 0.1

for i in range(maxiter):
    
    d = -F.grad(y)
    
    res_norm = jnp.linalg.norm(d)
    
    if res_norm < tol:
        break
    elif i % 100 == 0:
        print(f"iteration {i}, ||grad(F)|| = {res_norm}")
     
    slope = -jnp.linalg.norm(d)**2
        
    alpha = armijo(F, y, d, slope, alpha0=1.2*alpha)
    
    x_old = x
    #print(f"F(y) = {F(y)}")
    #print(f'alpha = {alpha}')
    x = y + alpha*d
    #print(f"F(x) = {F(x)}")
    y = x + beta*(x - x_old)
    #print(f"F(y) = {F(y)}")
    #print()
    
F.print_info()
```

```python
# Nesterov constant
x = jnp.array([1.,1.])
y = x.copy()

F.reset()

alpha = 1
beta = 0.1
lmb = 0

for i in range(maxiter):
    
    d = -F.grad(y)
    
    res_norm = jnp.linalg.norm(d)
    
    if res_norm < tol:
        break
    elif i % 100 == 0:
        print(f"iteration {i}, ||grad(F)|| = {res_norm}")

    alpha = 1/80
    
    x_old = x
    #print(f"F(y) = {F(y)}")
    #print(f'alpha = {alpha}')
    x = y + alpha*d
    
    
    
    lmb_old = lmb
    lmb = 0.5*(1+jnp.sqrt(1+4*lmb**2))
    beta = (lmb_old-1)/lmb
    
    
    #print(f"F(x) = {F(x)}")
    y = x + beta*(x - x_old)
    #print(f"F(y) = {F(y)}")
    #print()
    
F.print_info()
```

```python
#log vs log-log plots
import numpy as np
import matplotlib.pyplot as plt

k = np.arange(1.0, 101.0)

y1 = 1 / k
y2 = 1 / k**2

plt.figure(figsize=(10, 6))

plt.plot(k, y1, label='1/k', linewidth=2)
plt.plot(k, y2, label='1/k^2', linewidth=2)

#axes to a logarithmic scale 
plt.yscale('log')
#plt.xscale('log')

plt.xlabel('Iteration (k)', fontsize=12)
plt.ylabel('Convergence Rate', fontsize=12)
plt.title('Plot of Convergence Rates: 1/k vs 1/k^2', fontsize=14)
plt.legend(fontsize=12)
#plt.grid(True, which="both", ls="--", alpha=0.7)

plt.show()
```
