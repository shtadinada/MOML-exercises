# Step size selection for the gradient descent method

```python
from jax import grad, jvp, jit
import jax.numpy as jnp
```

```python
class Tracer():
    
    def __init__(self, f):
        self._f = jit(f)
        self._grad = jit(grad(f))
        
        self.counter_f = 0
        self.counter_grad_f = 0
        
    def __call__(self, x):
        self.counter_f += 1
        return self._f(x)
    
    def grad(self, x):
        self.counter_grad_f += 1
        return self._grad(x)
    
    def reset(self):
        self.counter_f = 0
        self.counter_grad_f = 0
    
    def print_info(self):
        print(f"evals: {self.counter_f}\n"
              f"grad evals: {self.counter_grad_f}")
```

```python
a = 20

F = Tracer(lambda x: a*x[0]**2 + x[1]**2)

F(jnp.array([1.,1.]))

F.print_info()

F.grad(jnp.array([1.,1.]))

F.print_info()
```
