# Overview
In this notebook, we will build and train neural networks using [JAX](https://jax.readthedocs.io/). We will consider the following examples:

- MNIST dataset of handwritten digits
- Image reconstruction example

```python
import numpy as np
import jax
import jax.numpy as jnp
from jax import random, grad
import optax
from keras.datasets import mnist
from PIL import Image
import matplotlib.pyplot as plt
from skimage import data
```

## MNIST example of handwritten digits
The MNIST dataset is a standard machine learning benchmark containing 70,000 grayscale 28x28 pixel images of handwritten digits from zero to nine. 
It can be loaded directly through most deep learning libraries like `keras.datasets` or downloaded manually from [Yann LeCun's official website](http://yann.lecun.com/).

- 60,000 for training
- 10,000 for testing

```python
# Load MNIST data
(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train.reshape(-1, 28*28) / 255.0
x_test  = x_test.reshape(-1, 28*28) / 255.0

y_train = jax.nn.one_hot(y_train, 10)
y_test  = jax.nn.one_hot(y_test, 10)

x_train = jnp.array(x_train)
y_train = jnp.array(y_train)
x_test  = jnp.array(x_test)
y_test  = jnp.array(y_test)

# Initialization 
def init_mnist_params(key):
    return {
        "W1": random.normal(key, (784, 128)) * 0.01,
        "b1": jnp.zeros(128),
        "W2": random.normal(key, (128, 10)) * 0.01,
        "b2": jnp.zeros(10),
    }

def mnist_forward(params, x):
    # can change activation function
    h = jax.nn.relu(x @ params["W1"] + params["b1"])
    return h @ params["W2"] + params["b2"]

def mnist_loss(params, x, y):
    logits = mnist_forward(params, x)
    return optax.softmax_cross_entropy(logits, y).mean()

# Accuracy method
def mnist_accuracy(params, x, y):
    logits = mnist_forward(params, x)
    preds = jnp.argmax(logits, axis=1)
    labels = jnp.argmax(y, axis=1)
    correct = jnp.sum(preds == labels)
    accuracy = correct / x.shape[0]
    return int(correct), float(accuracy)
```

```python
# Incremental GD
def train_incremental(params, lr=0.01, steps=20000):
    losses = []
    n_samples = x_train.shape[0]
    
    @jax.jit
    def update(p, x_i, y_i):
        # We add a batch dimension so shapes are (1, 784) and (1, 10)
        x_i = x_i.reshape(1, -1)
        y_i = y_i.reshape(1, -1)
        
        g = grad(mnist_loss)(p, x_i, y_i)
        p = jax.tree_util.tree_map(lambda w, dw: w - lr * dw, p, g)
        return p, mnist_loss(p, x_i, y_i)

    # randomly picking ω_i(SGD randomly pick 1 index)
    #indices = jax.random.randint(random.PRNGKey(37), shape=(steps,), minval=0, maxval=n_samples)

    for step in range(steps):
        idx = step % n_samples
        #idx = indices[step]    #SGD
        x_i = x_train[idx]
        y_i = y_train[idx]
        
        params, loss_val = update(params, x_i, y_i)
        losses.append(loss_val)

        if step % 2000 == 0:
            print(f"Iteration {step}, Loss: {loss_val:.4f}")
            
    return params, jnp.array(losses)


mnist_params = init_mnist_params(random.PRNGKey(1))

# Run Incremental GD
igd_params, igd_train_loss = train_incremental(mnist_params, lr=0.01, steps=20000)

# Compute accuracy
igd_correct, igd_acc = mnist_accuracy(igd_params, x_test, y_test)

print(f"\nFinal Test Accuracy:   {igd_correct} / {len(x_test)}   ({igd_acc*100:.2f}%)")
```

```python
# Training method (Gradient Descent)
    # can change lr and steps
def train_gd(params, lr=0.1, steps=200):
    losses = []
    #speeds up the full-batch gradient calculation
    @jax.jit
    def update(p, x, y):
        g = grad(mnist_loss)(p, x, y)
        p = jax.tree_util.tree_map(lambda w, dw: w - lr * dw, p, g)
        return p, mnist_loss(p, x, y)

    for step in range(steps):
        params, loss_val = update(params, x_train, y_train)
        losses.append(loss_val)

        if step % 50 == 0:
            print(f"Iteration {step}, Loss: {loss_val:.4f}")
        
    return params, jnp.array(losses)

# Run Gradient descent
mnist_params = init_mnist_params(random.PRNGKey(1))
gd_params, gd_train_loss = train_gd(mnist_params)

# Compute accuracy
gd_correct, gd_acc = mnist_accuracy(gd_params, x_test, y_test)

print(f"accuracy:   {gd_correct} / {len(x_test)}   ({gd_acc*100:.2f}%)")
```

```python
# Heavy ball
def train_heavy_ball(params, lr=0.1, beta=0.95, steps=200):
    losses = []
    
    # Initialize velocity to zeros with the exact same structure as params
    velocity = jax.tree_util.tree_map(jnp.zeros_like, params)
    
    @jax.jit
    def update(p, v, x, y):
        # compute gradient
        g = jax.grad(mnist_loss)(p, x, y)
        
        # v_{t+1} = beta * v_t + lr * g_t
        v = jax.tree_util.tree_map(lambda v_w, gw: beta * v_w + lr * gw, v, g)
        
        # w_{t+1} = w_t - v_{t+1}
        p = jax.tree_util.tree_map(lambda pw, v_w: pw - v_w, p, v)
        
        return p, v, mnist_loss(p, x, y)

    for step in range(steps):
        params, velocity, loss_val = update(params, velocity, x_train, y_train)
        losses.append(loss_val)

        if step % 50 == 0:
            print(f"Iteration {step}, Loss: {loss_val:.4f}")
            
    return params, jnp.array(losses)

# Run Heavy Ball
mnist_params = init_mnist_params(jax.random.PRNGKey(1))
hb_params, hb_train_loss = train_heavy_ball(mnist_params)

# Compute accuracy
hb_correct, hb_acc = mnist_accuracy(hb_params, x_test, y_test)

print(f"accuracy:   {hb_correct} / {len(x_test)} ({hb_acc*100:.2f}%)")
```

```python
# Beta values we want to compare
betas = [0.05, 0.1, 0.8, 0.95]

plt.figure(figsize=(10, 6))
base_params = init_mnist_params(jax.random.PRNGKey(1))

for beta in betas:
    print(f"\nTraining with beta = {beta}:")
    # Train 
    _, losses = train_heavy_ball(base_params, lr=0.1, beta=beta, steps=200)
    
    # Plot 
    plt.plot(losses, label=f"β = {beta}", linewidth=2)

plt.title("Heavy Ball Convergence across different Momentum values", fontsize=14)
plt.xlabel("Iteration", fontsize=12)
plt.ylabel("Training Loss", fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Armijo + slope checks are expensive!
import time
from jax.flatten_util import ravel_pytree

def armijo(F, x, d, slope, c=0.5, tau=0.5, alpha0=1):
    
    alpha = alpha0
    Fx = F(x)
    cslope = c*slope
    
    while F(x + alpha*d) > Fx + alpha*cslope:
        
        alpha *= tau
        
    return alpha

try:
    keys = random.split(random.PRNGKey(0), 2)
    n_hidden = 40960  # change
    large_p = {
        "W1": random.normal(keys[0], (784, n_hidden)) * 0.01,
        "b1": jnp.zeros(n_hidden),
        "W2": random.normal(keys[1], (n_hidden, 10)) * 0.01,
        "b2": jnp.zeros(10),
    }
    v = jax.tree_util.tree_map(jnp.zeros_like, large_p)

    # Measure Standard Heavy Ball Step 
    t0 = time.perf_counter()
    g = jax.grad(mnist_loss)(large_p, x_train, y_train)
    v = jax.tree_util.tree_map(lambda v_w, gw: 0.9 * v_w + 0.1 * gw, v, g)
    p_std = jax.tree_util.tree_map(lambda pw, v_w: pw - v_w, large_p, v)
    jax.block_until_ready(p_std)  
    hb_time = time.perf_counter() - t0

    # Measure Slope Check + Armijo Line Search Overhead 
    t0 = time.perf_counter()
    x_flat, unravel = ravel_pytree(large_p)
    g_flat, _ = ravel_pytree(g)
    v_flat, _ = ravel_pytree(v)
    
    d = -v_flat
    slope = jnp.dot(g_flat, d).block_until_ready()

    # Define 1D wrapper F so F(x + alpha*d) works inside your armijo function
    F = lambda w: mnist_loss(unravel(w), x_train, y_train).block_until_ready()
    alpha = armijo(F, x_flat, d, slope)
    ls_time = time.perf_counter() - t0

    # Print Metrics
    param_mb = (g_flat.size * 4) / (1024**2)
    act_mb = (x_train.shape[0] * n_hidden * 4) / (1024**2)
    
    print(f"Parameter count:       {g_flat.size:,} ({param_mb:.1f} MB per state copy)")
    print(f"Extra memory overhead: {param_mb * 3:.1f} MB (for x_flat, g_flat, v_flat)")  
    print(f"Forward pass memory:   {act_mb:.1f} MB per F(x + alpha*d) evaluation")
    print(f"Standard HB step time: {hb_time:.4f} s")
    print(f"Slope + Armijo time:   +{ls_time:.4f} s (alpha={alpha}, {(hb_time + ls_time)/hb_time:.1f}x slower overall)")

except Exception as e:
    print("Line search failed due to memory/computations explosion:", e)
```

```python
# Nesterov
def train_nesterov(params, lr=0.1, beta=0.95, steps=200):
    losses = []
    
    # Initialize velocity to zeros
    velocity = jax.tree_util.tree_map(jnp.zeros_like, params)
    
    @jax.jit
    def update(p, v, x, y):
        # calculate the "lookahead" parameters: p_lookahead = p + beta * v
        p_lookahead = jax.tree_util.tree_map(lambda pw, v_w: pw + beta * v_w, p, v)
        
        # gradient at the lookahead point
        g = jax.grad(mnist_loss)(p_lookahead, x, y)
        
        # v_{t+1} = beta * v_t - lr * g_lookahead
        v = jax.tree_util.tree_map(lambda v_w, gw: beta * v_w - lr * gw, v, g)
        
        # w_{t+1} = w_t + v_{t+1}
        p = jax.tree_util.tree_map(lambda pw, v_w: pw + v_w, p, v)
        
        return p, v, mnist_loss(p, x, y)

    for step in range(steps):
        params, velocity, loss_val = update(params, velocity, x_train, y_train)
        losses.append(loss_val)

        if step % 50 == 0:
            print(f"Iteration {step}, Loss: {loss_val:.4f}")
            
    return params, jnp.array(losses)

# Run Nesterov
mnist_params = init_mnist_params(jax.random.PRNGKey(1))
nag_params, nag_train_loss = train_nesterov(mnist_params)

# Compute accuracy
nag_correct, nag_acc = mnist_accuracy(nag_params, x_test, y_test)

print(f"accuracy:   {nag_correct} / {len(x_test)} ({nag_acc*100:.2f}%)")
```

```python
plt.figure(figsize=(12, 7))

# Plot full-batch methods (these will end at x = 200)
plt.plot(gd_train_loss, label="Gradient Descent", linewidth=2)
plt.plot(hb_train_loss, label="Heavy Ball", linewidth=2)
plt.plot(nag_train_loss, label="Nesterov Accelerated Gradient", linewidth=2)

# Plot Incremental GD (this will end at x = 20000)
iterations = range(0, len(igd_train_loss), 100)
igd_train_loss_sampled = igd_train_loss[::100]
plt.plot(igd_train_loss_sampled, label="Incremental GD (SGD)", linewidth=1, alpha=0.6)

plt.title("Training Loss Comparison: MNIST Optimization Methods", fontsize=16)
plt.xlabel("Iterations (Steps)", fontsize=14)
plt.ylabel("Training Loss", fontsize=14)

# We use a log scale for the y-axis because the losses drop exponentially 
# and it makes it much easier to see which method performs best at the bottom.
plt.yscale('log')

plt.legend(fontsize=12)
plt.grid(True, linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

## Image Reconstruction: Memorizing with Neural Networks
In this example, we use the famous "Cameraman" dataset to demonstrate how neural networks can reconstruct visual data. Instead of feeding an image *through* the network, we train the network to *become* the image.

We treat the image as a physical coordinate grid. 
- **The Input:** We feed the network just two numbers: the (X, Y) space coordinates of a single pixel. 
- **The Output:** The network predicts the grayscale brightness for that exact location. 
- **The Reconstruction:** By guessing the brightness for every coordinate and correcting its mistakes via *optimization algorithms*, the original image is slowly memorized into the network's hidden layers. To reconstruct the final image, we simply ask the network to predict the color for every coordinate on a blank canvas.

```python
# Load the Cameraman
original_img = data.camera()

# Downsample it to 128x128 so it trains instantly
img = original_img[::4, ::4] / 255.0
H, W = img.shape

# Create the dataset: (X, Y) inputs -> Pixel brightness output
x_coords = np.linspace(-1, 1, W)
y_coords = np.linspace(-1, 1, H)
xx, yy = np.meshgrid(x_coords, y_coords)

# Flatten into lists of inputs (coordinates) and outputs (pixels)
X_data = jnp.array(np.c_[xx.ravel(), yy.ravel()], dtype=jnp.float32)
Y_data = jnp.array(img.ravel().reshape(-1, 1), dtype=jnp.float32)


# Initialization and model
def init_camera_params(key):
    # 4 hidden layers of 128 neurons
    keys = random.split(key, 5)
    return {
        "W1": random.normal(keys[0], (2, 128)) * 0.1,  "b1": jnp.zeros(128),
        "W2": random.normal(keys[1], (128, 128)) * 0.1, "b2": jnp.zeros(128),
        "W3": random.normal(keys[2], (128, 128)) * 0.1, "b3": jnp.zeros(128),
        "W4": random.normal(keys[3], (128, 128)) * 0.1, "b4": jnp.zeros(128),
        "W5": random.normal(keys[4], (128, 1)) * 0.1,   "b5": jnp.zeros(1),
    }

def camera_forward(params, x):
    h = jax.nn.relu(x @ params["W1"] + params["b1"])
    h = jax.nn.relu(h @ params["W2"] + params["b2"])
    h = jax.nn.relu(h @ params["W3"] + params["b3"])
    h = jax.nn.relu(h @ params["W4"] + params["b4"])
    # Use sigmoid at the end to ensure the output brightness is strictly between 0 and 1
    return jax.nn.sigmoid(h @ params["W5"] + params["b5"])

# Mean Squared Error for regression
def mse_loss(params, x, y):
    preds = camera_forward(params, x)
    return jnp.mean((preds - y) ** 2)


# Training
#def train_camera_gd(params, lr=0.5, steps=2000):
#    @jax.jit
#    def update(p, x, y):
#        loss_val, g = jax.value_and_grad(mse_loss)(p, x, y)
#        p = jax.tree_util.tree_map(lambda w, dw: w - lr * dw, p, g)
#        return p, loss_val

#    losses = []
#    for step in range(steps):
#        params, loss_val = update(params, X_data, Y_data)
#        if step % 500 == 0:
#            print(f"Iteration {step}, Loss: {loss_val:.4f}")
            
#    return params

# Heavy Ball Training
def train_camera_heavy_ball(params, lr=0.5, beta=0.95, steps=2000):
    
    # Initialize velocity to zeros with the exact same structure as params
    import jax.numpy as jnp # Ensure jnp is available for zeros_like
    velocity = jax.tree_util.tree_map(jnp.zeros_like, params)
    
    @jax.jit
    def update(p, v, x, y):
        loss_val, g = jax.value_and_grad(mse_loss)(p, x, y)
        v = jax.tree_util.tree_map(lambda v_w, gw: beta * v_w + lr * gw, v, g)
        p = jax.tree_util.tree_map(lambda pw, v_w: pw - v_w, p, v)
        
        return p, v, loss_val

    losses = []
    for step in range(steps):
        params, velocity, loss_val = update(params, velocity, X_data, Y_data)
        losses.append(loss_val)
        
        if step % 500 == 0:
            print(f"Iteration {step}, Loss: {loss_val:.4f}")
            
    return params, jnp.array(losses)

# Run it
cam_params = init_camera_params(random.PRNGKey(7))

trained_cam_params, cam_losses = train_camera_heavy_ball(cam_params, lr=0.5, beta=0.95, steps=2000)

# Visualize the Reconstruction
predicted_pixels = camera_forward(trained_cam_params, X_data)
reconstructed_img = predicted_pixels.reshape(H, W)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title("Original Image (128x128)")
axes[0].axis('off')

axes[1].imshow(reconstructed_img, cmap='gray')
axes[1].set_title("Neural Network Reconstruction")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
